```python
import json
import random
from collections import defaultdict
from pathlib import Path

# Set random seed for reproducibility
random.seed(42)

# Load the four-people testset
testset_path = Path('four-people_testset.json')
with open(testset_path, 'r') as f:
    testset = json.load(f)

print(f"Total images in testset: {len(testset)}")
print(f"Sample entry: {testset[0]}")
```

```python
# Group images by label
images_by_label = defaultdict(list)
for entry in testset:
    for label in entry['labels']:
        images_by_label[label].append(entry)

print(f"\nLabels found: {list(images_by_label.keys())}")
for label, images in images_by_label.items():
    print(f"  {label}: {len(images)} images")
```

```python
# Create training set: 60% from each label
training_set_60 = []
for label, images in images_by_label.items():
    num_to_select = max(1, int(len(images) * 0.6))
    selected_images = random.sample(images, num_to_select)
    training_set_60.extend(selected_images)
    print(f"{label}: selected {num_to_select}/{len(images)} images")

print(f"\nTotal images in 60% training set: {len(training_set_60)}")

# Save the 60% training set
output_path_60 = Path('four-people-trainset.json')
# with open(output_path_60, 'w') as f:
    # json.dump(training_set_60, f, indent=2)

# print(f"Saved to {output_path_60}")
```

```python
# Create sample set: 20% from the 60% training set
num_to_select_sample = max(1, int(len(training_set_60) * 0.2))
training_set_sample = random.sample(training_set_60, num_to_select_sample)

print(f"Selected {num_to_select_sample} images (20% of 60% set) for sample training set")
print(f"Total images in sample training set: {len(training_set_sample)}")

# Save the sample training set
output_path_sample = Path('four-people-trainset-sample.json')
# with open(output_path_sample, 'w') as f:
#     json.dump(training_set_sample, f, indent=2)

# print(f"Saved to {output_path_sample}")
```

```python
# Create test set: remaining images NOT in the training set (no overlap)
training_set_paths = {entry['path'] for entry in training_set_60}

test_set = [entry for entry in testset if entry['path'] not in training_set_paths]

print(f"Training set size: {len(training_set_60)}")
print(f"Test set size: {len(test_set)}")
print(f"Total (should equal original): {len(training_set_60) + len(test_set)} (original: {len(testset)})")

# Verify no overlap
test_set_paths = {entry['path'] for entry in test_set}
overlap = training_set_paths & test_set_paths
print(f"\nOverlap between train and test: {len(overlap)} images")
assert len(overlap) == 0, "ERROR: There is overlap between training and test sets!"

# Show distribution per label
print("\nPer-label split:")
test_by_label = defaultdict(list)
for entry in test_set:
    for label in entry['labels']:
        test_by_label[label].append(entry)
for label in images_by_label:
    train_count = len([e for e in images_by_label[label] if e['path'] in training_set_paths])
    test_count = len(test_by_label.get(label, []))
    print(f"  {label}: train={train_count}, test={test_count}")

# Save the test set
output_path_test = Path('four-people-testset-holdout.json')
with open(output_path_test, 'w') as f:
    json.dump(test_set, f, indent=2)

print(f"\nSaved test set to {output_path_test}")
```
